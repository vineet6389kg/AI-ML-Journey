```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
print(arr)
```

```python
# create

arr1 = np.zeros((2,3), dtype = "int64")
print(arr1,arr1.shape)

arr1 = np.ones((2,3),)
print(arr1,arr1.shape)
```

```python
# operations on array

arr = np.array([[1,2,3],[4,5,6]])
print(arr, arr.shape)

reshaped = arr.reshape((3,2))
print(reshaped, reshaped.shape)
```
